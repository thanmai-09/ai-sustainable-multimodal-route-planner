## Sustainable Multimodal Mobility Network Dataset

## 1. Dataset Overview

This notebook creates a connected transportation network dataset for the
AI-Powered Sustainable Multimodal Route Planner.

The dataset represents connections between locations and provides transportation
options such as Walking, Bicycle, Public Transport, and Car.

Each connection contains information about distance, travel time, cost, and CO₂
emissions. The resulting dataset will be used to generate and evaluate
multimodal journeys based on user-defined time and budget constraints.
